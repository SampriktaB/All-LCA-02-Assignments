```python
import pandas as pd

# 1. Read CSV File
df = pd.read_csv(r"C:\Samprikta\Students.csv")
print("Original Data:")
print(df)

# 2. Display First 5 Rows
print("\nFirst 5 Rows:")
print(df.head())

# 3. Display Information
print("\nDataset Information:")
df.info()

# 4. Display Statistics
print("\nStatistics:")
print(df.describe())

# 5. Check Missing Values
print("\nMissing Values:")
print(df.isna().sum())

# 6. Fill Missing Values
df["MARKS"] = df["MARKS"].fillna(df["MARKS"].mean())

# 7. Remove Duplicates
df = df.drop_duplicates()

# 8. Drop Irrelevant Column
df = df.drop(columns=["AGE"])

# 9. Sort By Marks
df = df.sort_values(by="MARKS", ascending=False)

# 10. Filter Students With Marks Above 85
print("\nStudents With Marks Above 85:")
print(df.loc[df["MARKS"] > 85])

# 11. Display Cleaned Data
print("\nCleaned Data:")
print(df)

# 12. Save Cleaned Data
df.to_csv("Cleaned_Students.csv", index=False)
```
