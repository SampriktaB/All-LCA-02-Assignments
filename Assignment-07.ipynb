```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load the CSV dataset
df = pd.read_csv(r"C:\Samprikta\Assignment-07-Dataset.csv")
```

```python
# Display the dataset
print("Dataset:")
print(df)
```

```python
# Line Plot
plt.figure(figsize=(7, 4.5))
plt.plot(df["Student"], df["Python"], marker="o")
plt.title("Python Marks of Students")
plt.xlabel("Student")
plt.ylabel("Python Marks")
plt.grid(True)
plt.show()
```

```python
# Bar Plot
plt.figure(figsize=(7, 4.5))
plt.bar(df["Student"], df["Maths"])
plt.title("Maths Marks of Students")
plt.xlabel("Student")
plt.ylabel("Maths Marks")
plt.show()
```

```python
# Scatter Plot
plt.figure(figsize=(7, 4.5))
plt.scatter(df["Maths"], df["Python"], s=70)
plt.title("Maths vs Python Marks")
plt.xlabel("Maths Marks")
plt.ylabel("Python Marks")
plt.grid(True)
plt.show()
```

```python
# Correlation Matrix and Heatmap
correlation = df[["Maths", "Python", "Attendance"]].corr()
print("Correlation Matrix:")
print(correlation)
```

```python
plt.figure(figsize=(6.5, 5))
sns.heatmap(correlation, annot=True, cmap="Blues", vmin=-1, vmax=1)
plt.title("Correlation Heatmap")
plt.show()
```

```python
# Histogram of Python Marks
plt.figure(figsize=(7, 4.5))
plt.hist(df["Python"], bins=5)
plt.title("Histogram of Python Marks")
plt.xlabel("Python Marks")
plt.ylabel("Number of Students")
plt.show()
```

```python
# Box Plot of Maths and Python Marks
plt.figure(figsize=(7, 4.5))
plt.boxplot([df["Maths"], df["Python"]], tick_labels=["Maths", "Python"])
plt.title("Box Plot of Maths and Python Marks")
plt.ylabel("Marks")
plt.show()
```

```python
# Pie Chart of Students from Each City
city_count = df["City"].value_counts()
plt.figure(figsize=(7, 4.5))
plt.pie(city_count, labels=city_count.index, autopct="%1.1f%%")
plt.title("Students from Each City")
plt.show()
```

```python
# Calculate and Display Average Marks and Attendance
print("Average Maths Marks:",
df["Maths"].mean())
print("Average Python Marks:",
df["Python"].mean())
print("Average Attendance:",
df["Attendance"].mean())
```
